##### Creation of Spark Session

```python
from pyspark.sql import SparkSession
##### Create a new spark session
session = SparkSession.builder.appName('Dataframe').getOrCreate()
##### get the version of spark session
print(session.version)
```

2.  Create a DataFrame using an explicitly defined StructType schema

```python
from pyspark.sql.functions import *
from pyspark.sql.types     import StructType, StructField, StringType, IntegerType, FloatType

##### Create the dataframe using explicitly defined struct type

##### Create the data
employees = [
    (101, "Rahul", "IT", 60000),
    (102, "Priya", "HR", 45000),
    (103, "Amit", "IT", 75000),
    (104, "Sneha", "Finance", 55000),
    (105, "Raj", "IT", 60000),
    (106, "Neha", "HR", None),
    (107, "Karan", "Finance", 80000),
    (108, "Anjali", "IT", 70000),
    (109, "Vikram", "Sales", 50000),
    (110, "Pooja", "HR", 48000),
    (111, "Arjun", "Finance", 90000),
    (112, "Meera", "IT", None),
    (113, "Rohit", "Sales", 65000),
    (114, "Kavya", "Finance", 72000),
    (115, "Suresh", "IT", 85000),
    (116, "Divya", "HR", 52000),
    (117, "Manoj", "Sales", None),
    (118, "Isha", "IT", 95000),
    (119, "Nitin", "Finance", 62000),
    (120, "Riya", "HR", 58000)
]
##### Create the schema
schema = StructType([
    StructField('emp_id', IntegerType(), True)                  ,
    StructField('emp_name', StringType(), True)                 ,
    StructField('department', StringType(), True)               ,
    StructField('salary', FloatType(),  True)
])

##### Create the dataframe using explicitly defined schema
df = session.createDataFrame(employees, schema)

##### Display the dataframe
df.show()
```

14.  Create a DataFrame with an ArrayType column

```python
from pyspark.sql.types import IntegerType, StringType, FloatType, ArrayType, StructType, StructField

##### define the data
employees = [
    (101, "Rahul", ["Python", "SQL", "PySpark"]),
    (102, "Priya", ["Excel", "Power BI"]),
    (103, "Amit", ["Python", "Azure", "Databricks"]),
    (104, "Sneha", ["SQL", "ETL"]),
    (105, "Raj", ["Python", "SQL"])
]

##### define the schema
schema = StructType([
    StructField("emp_id"    ,   IntegerType()   ,   True)                       ,
    StructField("emp_name"  ,   StringType()    ,   True)                       ,
    StructField("skills"    ,   ArrayType(StringType()) ,   True)
])

##### define the dataframe
df = session.createDataFrame(employees, schema)

##### Display the dataframe
df.show()
```

15. Create a DataFrame with nested StructType columns

```python
from pyspark.sql.types import StringType, StructField, IntegerType, StringType, FloatType

# Employee data with nested address
employees = [
    (101, "Rahul", ("Hyderabad", "Telangana", 500001)),
    (102, "Priya", ("Bengaluru", "Karnataka", 560001)),
    (103, "Amit", ("Chennai", "Tamil Nadu", 600001)),
    (104, "Sneha", ("Pune", "Maharashtra", 411001)),
    (105, "Raj", ("Delhi", "Delhi", 110001))
]



##### Create the schema
schema = StructType([
    StructField("id", IntegerType(), True)  ,
    StructField("name", StringType(), True) ,
    StructField("address", StructType([
        StructField("city" ,  StringType()  ,   True)   ,
        StructField("state",  StringType()  ,   True)   ,
        StructField("pincode", IntegerType(),   True)
    ]))
])

##### Construct the dataframe from the schema
df = session.createDataFrame(employees, schema)

##### Display the dataframe
df.show()
```

16.  Create a DataFrame with a MapType column and display its schema

```python
from pyspark.sql.types import StringType, StructField, IntegerType, StringType, MapType

##### Create the dataset
employees = [
    (101, "Rahul", {"Python": 5, "SQL": 4, "PySpark": 3}),
    (102, "Priya", {"Excel": 3, "Power BI": 2}),
    (103, "Amit", {"Python": 4, "Azure": 3}),
    (104, "Sneha", {"SQL": 5, "ETL": 4}),
    (105, "Raj", {"Python": 3, "SQL": 2})
]

##### define the custom schema
schema = StructType([
    StructField('emp_id'    ,   IntegerType()   ,   True)               ,
    StructField('emp_name'  ,   StringType()    ,   True)               ,
    StructField('skills'    ,   MapType(StringType(), IntegerType())    ,   True)
])

##### Construct the dataframe
df = session.createDataFrame(employees, schema)

##### Display the dataframe
df.show()

##### Print the schema of the dataframe
df.printSchema()
```

17.  Create a PySpark DataFrame containing employee details and joining dates

```python
from pyspark.sql.types import StringType, IntegerType, StructType, StructField, DateType
from datetime import datetime

##### Create the employee data
employees = [
    (101, "Rahul", "2022-01-10"),
    (102, "Priya", "2021-05-15"),
    (103, "Amit", "2020-03-20"),
    (104, "Sneha", "2023-07-01"),
    (105, "Raj", "2022-08-12")
]

##### convert the date strings into Python date
employees = [
    (emp_id, emp_name, datetime.strptime(joining_date, "%Y-%m-%d"))
    for emp_id, emp_name, joining_date in employees
]

##### Create the schema for the above data
schema = StructType([
    StructField('emp_id'    ,   IntegerType()   ,   True)           ,
    StructField('emp_name'  ,   StringType()    ,   True)           ,
    StructField('joining_date', DateType()      ,   True)
])

##### Create the dataframe
df = session.createDataFrame(employees, schema)
##### Display the dataframe
df.show()
```
